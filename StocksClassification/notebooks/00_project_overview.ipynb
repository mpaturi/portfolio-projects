# Project overview – 5-year stock classification (AAPL, AMZN, MSFT, VOO, SPY)

Goal: Analyze 5 years of daily OHLCV data for AAPL, AMZN, MSFT, VOO, and SPY and build a simple classification model to predict whether the next-day return is positive (up) or not (down).

## Key questions

- How do the return and volatility profiles of AAPL, AMZN, and MSFT compare to broad S&P 500 ETFs (VOO, SPY)?
- Is there any exploitable signal in simple lagged returns and technical-style features for predicting up vs. down next day?
- How stable is model performance over time and across tickers?

## Assumptions / constraints

- Use only daily data and simple features (lagged returns, moving averages, rolling volatility, maybe ETF context features).
- No intraday data, no complex deep learning; keep everything interpretable and data-analyst-friendly.

2. Cleaning & alignment

4. Feature engineering
5. Modeling & evaluation
6. Visualization & communication
